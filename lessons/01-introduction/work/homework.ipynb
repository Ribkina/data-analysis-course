## Основной набор decisions.csv

```python
DATA = Path("../data")

assert DATA.is_dir(), "Не найдена папка data"

file = DATA / "decisions.csv"

assert file.exists(), f"Файл не найден: {file.resolve()}"

df = pd.read_csv(file)

print("Размер таблицы:", df.shape)

df.head()
```

## H1. Доля отказов

```python

total = len(df)

refused = (df["action_taken"] == 3).sum()
approved = df["action_taken"].isin([1, 2]).sum()

refused_share = refused / total
approved_share = approved / total

print("Всего записей:", total)
print("Отказов:", refused)
print("Доля отказов:", round(refused_share * 100, 2), "%")
print("Одобрений:", approved)
print("Доля одобрений:", round(approved_share * 100, 2), "%")

assert refused + approved == total
```

## H2. Одобрение без принятия предложения

```python
not_accepted = (df["action_taken"] == 2).sum()
issued = (df["action_taken"] == 1).sum()

not_accepted_share = not_accepted / total
issued_share = issued / total

difference = (issued_share - not_accepted_share) * 100

print("Одобрено, но не принято:", not_accepted)
print("Доля:", round(not_accepted_share * 100, 2), "%")
print("Выдано кредитов:", issued)
print("Доля выдач:", round(issued_share * 100, 2), "%")
print("Разница долей:", round(difference, 2), "п.п.")

assert not_accepted + issued == approved
```
